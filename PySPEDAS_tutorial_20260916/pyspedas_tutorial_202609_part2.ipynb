### 時系列データの四則演算について
- 時系列データの四則演算はpyspedas.add('var_name1', 'var_name2'), pyspedas.subtract('var_name1', 'var_name2'), pyspedas.multiply('var_name1', 'var_name2'), pyspedas.divide('var_name1', 'var_name2') でそれぞれ計算できます。subtract, divideの計算順はそれぞれ、'var_name1' - 'var_name2', 'var_name1' / 'var_name2' となります。
- pyspedasにはフィルター処理の関数が見当たらないので、たとえば、ハイパス処理は移動平均を差し引くなど、自分で計算する必要があります。(PySPEDASの関数を使っても、Pythonのnumpyやscipyの関数を使っても構いません。)
- ここでは、pyspedas.geopack.t89のtt89という関数を使って地磁気モデルの磁場を計算し、観測された磁場データからの差分(磁場の擾乱)をpyspedas.subtractを使って計算してみましょう。

```python
# Delete chaches
del_data('*')

# Import necessary modules
from ergpyspedas.erg import mgf, orb
from pyspedas import tplot, options, tplot_options, timespan, tplot_names, del_data, get_data, store_data, subtract
from pyspedas.geopack import ttrace2endpoint
from pyspedas.geopack.t89 import tt89
import numpy as np

# Setting
trange  = ['2023-04-24','2023-04-24/12:00']
timespan('2023-04-24',0.5,keyword='days')

# Load data
var_orb = orb(trange=trange)
var_mgf = mgf(trange=trange)

# Calculation
Bmodel  = tt89("erg_orb_l2_pos_gsm",
               iopt=1,   # ipot: 1:Kp=0,0+, 2:Kp=1-,1,1+, ...
               units_in="Re",   # unit of the orbit data ("km" or "Re")
               coord_in="GSM",   # input coordinate 
               coord_out="GSM",   # output coordinate
               igrf_only=False)   # If true, only the IGRF model is used to calculate the magnetic field.
subtract('erg_mgf_l2_mag_8sec_gsm','erg_orb_l2_pos_gsm_bt89', newname='dif_Bgsm')
```

```python
# Plot the calculated model field
# Setting
# Observational values
options('erg_mgf_l2_mag_8sec_gsm','yrange',[-600,600])
options('erg_mgf_l2_mag_8sec_gsm','ytitle','B(GSM,Arase)')

# Model
options('erg_orb_l2_pos_gsm_bt89','yrange',[-600,600])
options('erg_orb_l2_pos_gsm_bt89','ytitle','B(GSM,T89)')
options('erg_orb_l2_pos_gsm_bt89','ysubtitle','[nT]')
options('erg_orb_l2_pos_gsm_bt89','labels',['Bx','By','Bz'])

# Difference
options('dif_Bgsm','yrange',[-200,200])
options('dif_Bgsm','ytitle','\u0394B(GSM)')

# Plot
tplot(['erg_mgf_l2_mag_8sec_gsm','erg_orb_l2_pos_gsm_bt89','dif_Bgsm'])
pyspedas.tplot_names(['erg_mgf_l2_mag_8sec_gsm','erg_orb_l2_pos_gsm_bt89','dif_Bgsm'])
```

### 周波数解析
- 時系列データの周波数解析ではFFT(高速フーリエ変換)を用いて周波数スペクトルを計算することが常套手段です。
- 特に、pyspedas.tdpwrspc関数を使うと時系列データの時間-周波数スペクトルを計算することができます。

```python
# Delete chaches
del_data('*')

# Setting
trange  = ['2023-04-24','2023-04-24/12:00']
timespan('2023-04-24',12,keyword='hours')

# Load data
var_orb = orb(trange=trange)
var_mgf = mgf(trange=trange)

# Calculation
Bmodel  = tt89("erg_orb_l2_pos_gsm",
               iopt=1,   # ipot: 1:Kp=0,0+, 2:Kp=1-,1,1+, ...
               units_in="Re",   # unit of the orbit data ("km" or "Re")
               coord_in="GSM",   # input coordinate 
               coord_out="GSM",   # output coordinate
               igrf_only=False)   # If true, only the IGRF model is used to calculate the magnetic field.
subtract('erg_mgf_l2_mag_8sec_gsm','erg_orb_l2_pos_gsm_bt89', newname='dif_Bgsm')

var_fft = pyspedas.tdpwrspc('dif_Bgsm',nboxpoints=256,nshiftpoints=64,bin=1)

# Setting for plot
options('dif_Bgsm','yrange', [-200, 100])
options('*_dpwrspc','colormap', 'viridis')
options(['dif_Bgsm_x_dpwrspc','dif_Bgsm_y_dpwrspc','dif_Bgsm_z_dpwrspc'],'ylog',0)
options(['dif_Bgsm_x_dpwrspc','dif_Bgsm_y_dpwrspc','dif_Bgsm_z_dpwrspc'],'zrange',[1.0e0,1.0e4])

options(['dif_Bgsm_x_dpwrspc','dif_Bgsm_y_dpwrspc','dif_Bgsm_z_dpwrspc'],'yrange',[0,0.03])
options('dif_Bgsm_x_dpwrspc','ytitle', r'$\Delta$ Bx(SM)')
options('dif_Bgsm_y_dpwrspc','ytitle', r'$\Delta$ By(SM)')
options('dif_Bgsm_z_dpwrspc','ytitle', r'$\Delta$ Bz(SM)')

# Plot
tplot(['dif_Bgsm',
       # 'dif_Bgsm_x_dpwrspc',
       # 'dif_Bgsm_y_dpwrspc',
       'dif_Bgsm_z_dpwrspc'], ysize = 6)
```

### 便利関数(2) 座標の変換 (field-aligned coordinates, minimum variance coordinates, LMN coordinates)

- Field-aligned coordinatesへの座標変換は`pyspedas.fac_matrix_make` と `pyspedas.tvector_rotate`を使うと手軽にできる。
- MVA座標系は`pyspedas.mva_matrix_make`、LMN座標系は `pyspedas.lmn_matrix_make`を使うと手軽に変換できる。

<u>Tips</u><br> NaNがあると `pyspedas.tvector_rotate`はエラーを引き起こすので、データの前処理でNaNを取り除く必要がある。

```python
#Import necessary modules
from ergpyspedas.erg import mgf, orb
from pyspedas import tplot, options, tplot_options, timespan, tplot_names, del_data, get_data, store_data, time_double
import numpy as np

# Setting
trange  = ['2023-04-24','2023-04-25']
timespan('2023-04-24/21:50:00',2,keyword='hours')

# Load data
var_orb = orb(trange=trange)
var_mgf = mgf(trange=trange)

# Create tplot variable of averaged magnetic fields without NaN
pyspedas.avg_data('erg_mgf_l2_mag_8sec_gse',res=300) #300-sec average
pyspedas.tinterpol('erg_mgf_l2_mag_8sec_gse-avg','erg_mgf_l2_mag_8sec_gse') # moving average is more appropriate for the reference field
dat  = pyspedas.get_data('erg_mgf_l2_mag_8sec_gse-avg-itrp')
mag  = dat.y
time = dat.times
pyspedas.store_data('mag_finite',data={'x':time[~np.isnan(mag[:,0])],'y':mag[~np.isnan(mag[:,0]),:]})
pyspedas.tplot_tools.data_att_getters_setters.set_any('mag_finite','coord_sys','GSE') # add metadata of the coordinate system of the magnetic field data

# Create a rotation matrix
pyspedas.fac_matrix_make('mag_finite', pos_var_name='erg_orb_l2_pos_gse')

# Rotate a vactor with the matrix
pyspedas.tvector_rotate( 'mag_finite_fac_mat','erg_mgf_l2_mag_8sec_gse')
```

 - `pyspedas.tinterp('var1','var2')`: データの補間をする関数。 'var2'を'var1'の時刻幅で補間する。出力は 'var1_tinterp'であることに注意。
