# Installation & Setup

## Install `uv`

- For macOS and Linux, you can install `uv` using the following command:
- macOS/Linux(WSL)の場合、ターミナルで以下のコマンドを実行し `uv`をインストールできます。
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

- For Windows, you can install `uv` using the following command:
- Windowsの場合、Powershellで以下のコマンドを実行し `uv`をインストールできます。
```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
``` 


## Setup a virtual environment for PySPEDAS
- To set up an environment for PySPEDAS, you can use the following command:
- PySPEDASの仮想環境をセットアップするには、以下のコマンドを使用できます。
```bash
uv init <name_for_environment> -p 3.13
```
- <name_for_environment> should be replaced with the desired name for your virtual environment. 
- <name_for_environment> は、使用したい仮想環境の名前に書き換えてください。 

- `-p 3.13` specifies that Python version 3.13 should be used for the environment. 
- `-p 3.13` は、仮想環境にPythonバージョン3.13を使用することを指定します。

For example, if you want to name your environment "pyspeadas_tutorial", you would run:
```bash
uv init pyspeadas_tutorial -p 3.13
```

- You can see the following output in the commandline
- 以下の出力がコマンドラインに表示されます。
```
Initialized project `pyspedas-tutorial` is loacated at `<path_to_environment>/pyspedas_tutorial`
```

## Setup PySPEDAS
- To install necessary packages for PySPEDAS, you can use the following command:
- PySPEDASに必要なパッケージをインストールします。

```bash
cd <path_to_environment>/pyspedas_tutorial
uv add ipykernel
uv add pyspedas
```

### installing plug-ins for the ERG spacecraft
- To install the plug-ins for the ERG spacecraft, you can use the following command:
- ERG衛星のプラグインをインストールするには、以下のコマンドを使用できます。
```bash
uv add https://github.com/ergsc-devel/pyspedas_plugin.git
```

- 使用感としては、`pip install`コマンドが`uv add`コマンドに置き換わったイメージです。


## An example for PySPEDAS
まずは時系列データのロード+プロットから始めてみましょう。
以下のコードを実行することで、PySPEDASを使った時系列データのプロットが可能です。

````python
# import necessary modules. 必要なモジュールのインポート
import pyspedas                 # import pyspedas modules
import ergpyspedas              # import modules of the bleeding-edge of the ERG-SC plugin

from pyspedas import tplot # shortcut for pyspedas.tplot function
import matplotlib.pyplot as plt # shortcut for matplotlib.pyplot module
import numpy as np # shortcut for numpy module

# # show the path to the imported PySPEDAS package
# print(pyspedas.__file__)
# print(ergpyspedas.__file__)
````

````python
# import a function of mgf in (package directory)/pyspedas/projects/erg/satellite/erg/mgf/mgf.py
from ergpyspedas.erg import mgf # (or from pyspedas.project.erg import mgf)

# set a range of dataset
trange = ['2023-04-23','2023-04-25'] # trange is a list type specifying the period for data loading

# Load mgf data
var_mgf = mgf(level='l2',datatype='8sec',coord='',trange=trange) #var_mgf contains names of loaded tplot-variables of MGF 
````

````python
# Show tplot variables
pyspedas.tplot_names()
````

````python
# Make a quick look plot for the mgf data
pyspedas.tplot([var_mgf[5], var_mgf[1]])
````

## Load ERG data as array format

````python
# Get data in the ndarray format
data_mgf_ndarray = pyspedas.get_data(var_mgf[4])

print(data_mgf_ndarray)
print()
print("Time: ", data_mgf_ndarray.times)
print("Size: ", data_mgf_ndarray.times.size)
print()
print("Data: ", data_mgf_ndarray.y)
print("Shape: ", data_mgf_ndarray.y.shape)
print()

# Get metadata 
data_mgf_gsm_meta = pyspedas.get_data(var_mgf[3], metadata=True)
print(data_mgf_gsm_meta.keys())
for key in data_mgf_gsm_meta['CDF'].keys():
    print(key)
    if key == 'VATT':
        for subkey in data_mgf_gsm_meta['CDF']['VATT'].keys():
            print(subkey, ':', data_mgf_gsm_meta['CDF']['VATT'][subkey])
    if key == 'GATT':
        for subkey in data_mgf_gsm_meta['CDF']['GATT'].keys():
            print(subkey, ':', data_mgf_gsm_meta['CDF']['GATT'][subkey])
    if key == 'FILENAME':
        print('FILENAME', ':', data_mgf_gsm_meta['CDF']['FILENAME'])
    if key == 'LABELS':
        print('LABELS', ':', data_mgf_gsm_meta['CDF']['LABELS'])
for att in data_mgf_gsm_meta['data_att']:
    print(att, ':', data_mgf_gsm_meta['data_att'][att])
for option in data_mgf_gsm_meta['plot_options']:
    print(option, ':', data_mgf_gsm_meta['plot_options'][option])
print()
print()
````
